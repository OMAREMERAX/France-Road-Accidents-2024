```python
import pandas as pd
```

```python
vehicules = pd.read_csv(
    "data/vehicules-2024.csv",
    sep=";",
    encoding="utf-8"
)

print(vehicules.shape)
print(vehicules.head())
print(vehicules.dtypes)
```

```python
vehicules.isnull().sum().sort_values(ascending=False)
```

```python
vehicules["occutc"].value_counts(dropna=False).head(20)
```

```python
vehicules = vehicules.drop(columns=["occutc"])
```

```python
print(vehicules.shape)
print(vehicules.isnull().sum().sum())
```

```python
for col in ["senc", "catv", "obs", "obsm", "choc", "manv", "motor"]:
    print(f"\n--- {col} ---")
    print(vehicules[col].value_counts().head(20))
```

```python
print(vehicules.shape)
print(vehicules.isnull().sum().sum())
```

```python
vehicules.to_csv(
    "data/vehicules_2024_clean.csv",
    index=False,
    encoding="utf-8-sig"
)
```
