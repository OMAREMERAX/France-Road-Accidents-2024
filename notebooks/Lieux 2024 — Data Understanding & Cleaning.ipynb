```python
import pandas as pd

lieux = pd.read_csv(
    "data/lieux-2024.csv",
    sep=";",
    encoding="utf-8"
)

print(lieux.shape)
print(lieux.head())
print(lieux.dtypes)
```

```python
lieux.isnull().sum()
```

```python
lieux["lartpc"].value_counts(dropna=False)
```

```python
lieux["v2"].value_counts(dropna=False).head(20)
```

```python
lieux = lieux.drop(columns=["lartpc"])
```

```python
lieux.shape
```

```python
lieux["v2"] = lieux["v2"].fillna("Unknown")
```

```python
lieux["v2"].isnull().sum()
```

```python
lieux["voie"] = lieux["voie"].fillna("Unknown")
```

```python
lieux.isnull().sum()
```

```python
lieux.dtypes
```

```python
for col in ["nbv", "pr", "pr1", "larrout"]:
    print(f"\n--- {col} ---")
    print(lieux[col].value_counts().head(20))
```

```python
lieux["nbv"] = lieux["nbv"].replace("#VALEURMULTI", "Unknown")
```

```python
lieux["nbv"].value_counts().head(15)
```

```python
print(lieux.shape)
print(lieux.isnull().sum().sum())
print(lieux["nbv"].value_counts().head(15))
```

```python
lieux.to_csv(
    "data/lieux_2024_clean.csv",
    index=False,
    encoding="utf-8-sig"
)
```
