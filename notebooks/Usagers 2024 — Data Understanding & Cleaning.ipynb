```python
import pandas as pd
```

```python
usagers = pd.read_csv(
    "data/usagers-2024.csv",
    sep=";",
    encoding="utf-8"
)

print(usagers.shape)
print(usagers.head())
print(usagers.dtypes)
```

```python
usagers.isnull().sum().sort_values(ascending=False)
```

```python
usagers["an_nais"] = usagers["an_nais"].astype("Int64")
```

```python
usagers["an_nais"] = usagers["an_nais"].fillna(-1)
```

```python
usagers["an_nais"].isnull().sum()
```

```python
for col in ["catu", "grav", "sexe", "trajet", "secu1", "secu2", "secu3", "locp", "actp", "etatp"]:
    print(f"\n--- {col} ---")
    print(usagers[col].value_counts().head(15))
```

```python
usagers["id_usager"] = (
    usagers["id_usager"]
    .str.replace("\u00a0", "", regex=False)
    .str.replace(" ", "", regex=False)
)

usagers["id_vehicule"] = (
    usagers["id_vehicule"]
    .str.replace("\u00a0", "", regex=False)
    .str.replace(" ", "", regex=False)
)
```

```python
print(usagers[["id_usager", "id_vehicule"]].head())
```

```python
print(usagers.isnull().sum().sum())
print(usagers.shape)
```

```python
usagers.to_csv(
    "data/usagers_2024_clean.csv",
    index=False,
    encoding="utf-8-sig"
)
```
