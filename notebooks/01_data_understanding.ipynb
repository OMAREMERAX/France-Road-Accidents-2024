```python
lum_names = {
    1: "Daylight",
    2: "Twilight or dawn",
    3: "Night - public lighting on",
    4: "Night - public lighting off",
    5: "Night - no public lighting"
}

car_clean["Light_Condition"] = car_clean["lum"].map(lum_names)
```

```python
weather_names = {
    1: "Normal",
    2: "Light rain",
    3: "Heavy rain",
    4: "Snow or hail",
    5: "Fog or smoke",
    6: "Strong wind",
    7: "Dazzling weather",
    8: "Overcast",
    9: "Other"
}

car_clean["Weather_Condition"] = car_clean["atm"].map(weather_names)
```
